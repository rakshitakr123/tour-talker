```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import deque
```

```python
def visualize_graph(graph, title="Graph"):
    pos = nx.spring_layout(graph)
    nx.draw(graph, pos, with_labels=True, node_color='lightblue', font_weight='bold')
    plt.title(title)
    plt.show()
```

```python
def visualize_broadcast(graph, start_node):
    visited = set()
    queue = deque([start_node])
    visited.add(start_node)

    pos = nx.spring_layout(graph)
    visualize_graph(graph, title="Network Broadcasting using BFS - Initial State")

    while queue:
        current_node = queue.popleft()
        print(f"Broadcasting message to node {current_node}")

        if current_node in graph:
            for neighbor in graph.neighbors(current_node):
                if neighbor not in visited:
                    queue.append(neighbor)
                    visited.add(neighbor)


                    nx.draw(graph, pos, with_labels=True, node_color='lightblue', font_weight='bold',
                            node_size=700, nodelist=[current_node], edgelist=[(current_node, neighbor)], edge_color='red')
                    plt.title(f"Broadcasting message from {current_node} to {neighbor}")
                    plt.show()
```

```python
network = nx.Graph()
network.add_edge(1, 2)
network.add_edge(1, 3)
network.add_edge(2, 4)
network.add_edge(2, 5)
network.add_edge(3, 6)
network.add_edge(3, 7)

start_node = 1
print(f"Starting network broadcasting from node {start_node}")
visualize_broadcast(network, start_node)
```
